# EDA: Backblaze Hard Drive Data
Exploratory Data Analysis for the Backblaze Q1 2023 Dataset.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

%matplotlib inline
sns.set_theme(style='darkgrid')
```

## Load Processed TTF Data

```
data_path = '../data/processed/backblaze_ttf.csv'
if os.path.exists(data_path):
    df = pd.read_csv(data_path)
    print(f'Loaded {len(df)} records')
    display(df.head())
else:
    print('Processed dataset not found. Please run ttf_labels.py first.')
```

## Class Imbalance Check

```
if 'failure' in locals() or 'df' in locals():
    plt.figure(figsize=(6, 4))
    sns.countplot(data=df, x='failure')
    plt.title('Distribution of Failures (Highly Imbalanced)')
    plt.show()
```

## Feature Distributions (SMART attributes)

```
smart_cols = [c for c in df.columns if 'smart' in c]
if 'df' in locals() and len(smart_cols) > 0:
    df[smart_cols].hist(bins=50, figsize=(15, 10))
    plt.tight_layout()
    plt.show()
```
